#### Calculate VIF for Multicollinearity

##### Obesity

```python
# Encode columns
mapping = {
    'Insufficient_Weight': 0,
    'Normal_Weight': 1,
    'Overweight_Level_I': 2,
    'Overweight_Level_II': 3,
    'Obesity_Type_I': 4,
    'Obesity_Type_II': 5,
    'Obesity_Type_III': 6,
}

obesity['Obesity Level Encoded'] = obesity['Obesity Level'].map(mapping)
obesity.drop(columns = 'Obesity Level', inplace = True)
```

```python
# Add intercept column
datawithconst = sm.add_constant(obesity)
```

```python
# Calculate VIF for each variable
vif_data = pd.DataFrame()
vif_data["Variable"] = datawithconst.columns
vif_data["VIF"] = [
    variance_inflation_factor(datawithconst.values, i) 
    for i in range(datawithconst.shape[1])
]

print(vif_data.sort_values('VIF'))
```

##### Cervical Cancer

```python
# Calculate percentage of missing values per column
missing_pct = (cervical_cancer.isnull().sum() / len(cervical_cancer)) * 100

# Format as a clean DataFrame and display only columns that have missing values
missing_df = pd.DataFrame({'Missing %': missing_pct})
missing_df = missing_df[missing_df['Missing %'] > 0].sort_values(by='Missing %', ascending=False)

print(missing_df.round(2))
```

```python
# Save target column
cancer = cervical_cancer['Dx:Cancer']

# Drop collinear columns, columns with high % missing values, and target column
cervical_cancer.drop(columns = ['STDs: Time since first diagnosis', 
                                'STDs: Time since last diagnosis',
                                'STDs',
                                'STDs (number)',
                                'STDs: Number of diagnosis',
                                'STDs:condylomatosis',
                                'Dx:Cancer',
                                'STDs:AIDS',
                                'STDs:cervical condylomatosis'
                               ],
                     inplace = True)
```

```python
# Drop NaN rows
cervical_cancer = cervical_cancer.replace([np.inf, -np.inf], np.nan).dropna()

# Add intercept column
datawithconst = sm.add_constant(cervical_cancer)

# Calculate VIF for each variable
vif_data = pd.DataFrame()
vif_data["Variable"] = datawithconst.columns
vif_data["VIF"] = [
    variance_inflation_factor(datawithconst.values, i) 
    for i in range(datawithconst.shape[1])
]

print(vif_data.sort_values('VIF'))
```

```python
# Add target column back
cervical_cancer['Dx:Cancer'] = cancer
```

#### Linear Regression with Polynomial Terms

**Diabetes**

```python
Xd = diabetes_indicators.drop('Diabetes_012', axis = 1)
yd = diabetes_indicators['Diabetes_012']
```

```python
Xd_train, Xd_test, yd_train, yd_test = train_test_split(Xd, yd, test_size = 0.2, random_state = 42)
```

```python
for i in range(0, 5):
    model = make_pipeline(PolynomialFeatures(degree = i), LinearRegression())
    model.fit(Xd_train,yd_train)
    yd_pred = model.predict(Xd_test)
    # Evaluate model
    print(f"degree: {i} Precision: {mean_absolute_error(yd_test, yd_pred):.2f}")
```

**Cervical Cancer**

```python
Xc = cervical_cancer.drop('Dx:Cancer', axis = 1)
yc = cervical_cancer['Dx:Cancer']
```

```python
Xc_train, Xc_test, yc_train, yc_test = train_test_split(Xc, yc, test_size = 0.2, random_state = 42)
```

```python
for i in range(0, 5):
    model = make_pipeline(PolynomialFeatures(degree = i), LinearRegression())
    model.fit(Xc_train, yc_train)
    yc_pred = model.predict(Xc_test)
    # Evaluate model
    print(f"degree: {i} MAE: {mean_absolute_error(yc_test, yc_pred):.2f}")
```
